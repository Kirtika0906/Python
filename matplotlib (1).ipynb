```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print('Matplotlib version:', matplotlib.__version__)
print('NumPy version:', np.__version__)
```

```python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.show() # This creates a basic line plot
```

```python
x = np.linspace(0, 10, 50)

# Basic plot
plt.figure(figsize=(10, 6))
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic Plot with Legend')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Format string examples
plt.figure(figsize=(12, 4))
plt.plot(x, x, 'r-', label='linear')
plt.plot(x, x**2, 'g--', label='quadratic')
plt.plot(x, x**3, 'b:', label='cubic')
plt.plot(x, np.sqrt(x), 'm-.', label='square root')
plt.legend()
plt.title('Different Line Styles')
plt.show()
```

```python
# Detailed styling
plt.figure(figsize=(10, 6))
plt.plot(x, np.sin(x), color='purple', linestyle='--', linewidth=2, marker='o', markersize=5, markerfacecolor='yellow', label='sin(x)')
plt.plot(x, np.cos(x), color='#FF5733', linestyle='-', linewidth=3, marker='s', markersize=4, alpha=0.7, label='cos(x)')
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)
plt.title('Detailed Styling', fontsize=14, fontweight='bold')
plt.legend(loc='upper right', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.7)
plt.show()
```

```python

# Multiple line styles
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.figure(figsize=(12, 6))
plt.plot(x, y1, 'b-', linewidth=2, label='sin(x)')
plt.plot(x, y2, 'r--', linewidth=2, label='cos(x)')
plt.plot(x, y3, 'g:', linewidth=2, label='sin(x)*cos(x)')
plt.fill_between(x, y1, y2, alpha=0.2, color='purple', label='area between')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Line Plot with Filled Area')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
np.random.seed(42)

# Basic scatter
x = np.random.randn(50)
y = np.random.randn(50)
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', alpha=0.6, s=50)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Basic Scatter Plot')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x = np.random.randn(100)
y = np.random.randn(100)
colors = np.random.randint(1, 5, 100)
sizes = np.random.randint(20, 200, 100)
plt.figure(figsize=(8, 6))
scatter = plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='viridis', edgecolors='black', linewidth=0.5)
plt.colorbar(scatter, label='Group')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter with Color and Size')
plt.show()
```

```python
# Scatter with regression-like data
x = np.linspace(0, 10, 100)
y = 2*x + 5 + np.random.normal(0, 3, 100)
plt.figure(figsize=(8, 6))
plt.scatter(x, y, c=y, cmap='coolwarm', s=30, alpha=0.7)
plt.plot(x, 2*x + 5, 'r--', label='True line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter with Trend Line')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Bubble chart
categories = ['A', 'B', 'C', 'D', 'E']
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 30, 50, 40, 20])
size = np.array([100, 400, 900, 600, 250])
plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=size, c=size, cmap='plasma', alpha=0.6, edgecolors='black')
for i, cat in enumerate(categories):
    plt.annotate(cat, (x[i], y[i]), textcoords="offset points", xytext=(0, 10), ha='center')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Bubble Chart')
plt.show()
```

```python
# Vertical bar chart
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]
plt.figure(figsize=(8, 5))
bars = plt.bar(categories, values, color=['red', 'green', 'blue', 'orange', 'purple'], edgecolor='black', linewidth=1.5)
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Vertical Bar Chart')
plt.bar_label(bars, padding=3)
plt.show()
```

```python
# Horizontal bar chart
plt.figure(figsize=(8, 5))
bars = plt.barh(categories, values, color='steelblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Category')
plt.title('Horizontal Bar Chart')
plt.bar_label(bars, padding=3)
plt.show()
```

```python
# Grouped bar chart
x = np.arange(5)
men = [20, 35, 30, 35, 27]
women = [25, 32, 34, 20, 25]
width = 0.35
plt.figure(figsize=(8, 5))
plt.bar(x - width/2, men, width, label='Men', color='lightblue', edgecolor='black')
plt.bar(x + width/2, women, width, label='Women', color='lightpink', edgecolor='black')
plt.xlabel('Group')
plt.ylabel('Scores')
plt.title('Grouped Bar Chart')
plt.xticks(x, categories)
plt.legend()
plt.show()
```

```python
np.random.seed(42)
data = np.random.randn(1000)

# Basic histogram
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, color='steelblue', edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Basic Histogram')
plt.show()
```

```python
# Multiple histograms overlaid
data1 = np.random.normal(0, 1, 1000)
data2 = np.random.normal(2, 1.5, 1000)
plt.figure(figsize=(8, 5))
plt.hist(data1, bins=30, alpha=0.5, label='Data 1', color='blue')
plt.hist(data2, bins=30, alpha=0.5, label='Data 2', color='red')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Overlaid Histograms')
plt.legend()
plt.show()
```

```python
# Basic pie chart
labels = ['A', 'B', 'C', 'D', 'E']
sizes = [30, 25, 20, 15, 10]
plt.figure(figsize=(7, 7))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['red', 'green', 'blue', 'orange', 'purple'])
plt.title('Basic Pie Chart')
plt.show()
```

```python
# Exploded pie chart
explode = (0.1, 0, 0, 0, 0)
plt.figure(figsize=(7, 7))
plt.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title('Exploded Pie Chart')
plt.show()
```

```python
# Donut chart
plt.figure(figsize=(7, 7))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, wedgeprops=dict(width=0.5, edgecolor='w'))
plt.title('Donut Chart')
plt.show()
```

```python
# Simple subplots
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x))
plt.title('Sine')
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x))
plt.title('Cosine')
plt.subplot(2, 2, 3)
plt.plot(x, np.tan(x))
plt.title('Tangent')
plt.ylim(-5, 5)
plt.subplot(2, 2, 4)
plt.plot(x, np.exp(-x))
plt.title('Exponential Decay')
plt.tight_layout()
plt.show()
```

```python

# Using subplots() (recommended)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].plot(x, np.sin(x), color='red')
axes[0, 0].set_title('Sine')
axes[0, 1].plot(x, np.cos(x), color='blue')
axes[0, 1].set_title('Cosine')
axes[1, 0].plot(x, x**2, color='green')
axes[1, 0].set_title('Quadratic')
axes[1, 1].plot(x, np.log(x + 1), color='orange')
axes[1, 1].set_title('Logarithm')
fig.suptitle('Four Functions', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Box plot
np.random.seed(42)
data = [np.random.normal(0, std, 100) for std in range(1, 5)]
plt.figure(figsize=(8, 5))
plt.boxplot(data, labels=['A', 'B', 'C', 'D'], patch_artist=True, boxprops=dict(facecolor='lightblue'))
plt.xlabel('Group')
plt.ylabel('Value')
plt.title('Box Plot')
plt.show()
```

```python
# Violin plot
plt.figure(figsize=(8, 5))
plt.violinplot(data, showmeans=True, showmedians=True)
plt.xlabel('Group')
plt.ylabel('Value')
plt.title('Violin Plot')
plt.xticks([1, 2, 3, 4], ['A', 'B', 'C', 'D'])
plt.show()
```

```python
# Heatmap
data = np.random.randint(1, 100, (10, 10))
plt.figure(figsize=(8, 6))
im = plt.imshow(data, cmap='YlOrRd', aspect='auto')
plt.colorbar(im, label='Value')
plt.xlabel('Column')
plt.ylabel('Row')
plt.title('Heatmap')
# Add text annotations
for i in range(10):
    for j in range(10):
        plt.text(j, i, data[i, j], ha='center', va='center', color='black', fontsize=8)
plt.show()
```

```python
# 3D plot
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
X, Y = np.meshgrid(np.linspace(-5, 5, 50), np.linspace(-5, 5, 50))
Z = np.sin(np.sqrt(X**2 + Y**2))
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Surface Plot')
plt.show()
```
