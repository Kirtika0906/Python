<h2>5.Add a New Column</h2>

<h2>6.Calculate Average Marks</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(student)

average = df["Marks"].mean()

print("Average Marks =", average)
```

<h2>6a.Other Basic Statistics (min, max, sum, describe)</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Marks": [85, 90, 88, 70]
}

df = pd.DataFrame(student)

print("Minimum Marks =", df["Marks"].min())
print("Maximum Marks =", df["Marks"].max())
print("Total Marks =", df["Marks"].sum())
print("\nSummary Statistics")
print(df["Marks"].describe())
```

<h2>7.Sort Data by Marks</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Marks": [85, 90, 88, 70]
}

df = pd.DataFrame(student)

print(df.sort_values(by="Marks", ascending=False))
```

<h2>8.Rename Columns</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(student)

df = df.rename(columns={"Marks": "Score"})

print(df)
```

<h2>9.Handling Missing Values</h2>

```python
import pandas as pd
import numpy as np

student = {
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Marks": [85, np.nan, 88, 70]
}

df = pd.DataFrame(student)

print("Check for Missing Values")
print(df.isnull())

print("\nFill Missing Values with 0")
print(df.fillna(0))

print("\nDrop Rows with Missing Values")
print(df.dropna())
```

<h2>10.Group Data using groupby()</h2>

```python
import pandas as pd

record = {
    "Name": ["Ankit", "Amit", "Aishwarya", "Priyanka", "Priya", "Shaurya"],
    "Stream": ["Math", "Commerce", "Science", "Math", "Math", "Science"],
    "Percentage": [88, 92, 95, 70, 65, 78]
}

df = pd.DataFrame(record)

print("Average Percentage by Stream")
print(df.groupby("Stream")["Percentage"].mean())
```

<h2>12.Iterate Over Rows using iterrows()</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(student)

for index, row in df.iterrows():
    print(row["Name"], "scored", row["Marks"])
```

<h1>Complete Program for Pandas in One</h1>

```python
import pandas as pd

# Create data
student = {
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 88, 75]
}

# Create DataFrame
df = pd.DataFrame(student)

# Display Data
print("Student Data")
print(df)

# Display first two rows
print("\nFirst Two Rows")
print(df.head(2))

# Display student names
print("\nNames")
print(df["Name"])

# Filter students with marks above 80
print("\nStudents with Marks > 80")
print(df[df["Marks"] > 80])

# Average marks
print("\nAverage Marks")
print(df["Marks"].mean())

# Add Grade column
df["Grade"] = ["B", "A", "A", "C"]

print("\nUpdated Data")
print(df)
```

<h1>TASKS for Pandas</h1>

```python
import pandas as pd
from io import StringIO
record = {
  'Name': ['Ankit', 'Amit', 'Aishwarya', 'Priyanka', 'Priya', 'Shaurya' ],
  'Age': [21, 19, 20, 18, 17, 21],
  'Stream': ['Math', 'Commerce', 'Science', 'Math', 'Math', 'Science'],
  'Percentage': [88, 92, 95, 70, 65, 78]}
df=pd.DataFrame(record)
print(df)
```

```python
print("reading only math and commerce students")
print(df[df['Stream'].isin(['Math', 'Commerce'])])
```

```python
print("reading only rows not having science stream")
print(df[df['Stream']!='Science'])
```

```python
print("reading the rows where the stream is math and age is less than 20")
print(df[(df['Stream']=='Math') & (df['Age']<20)])
```

```python
print("reading students with percentage above 80, sorted highest to lowest")
print(df[df['Percentage'] > 80].sort_values(by='Percentage', ascending=False))
```

```python
print("average percentage grouped by stream")
print(df.groupby('Stream')['Percentage'].mean())
```

```python
print("adding a Result column based on Percentage (Pass if >= 70, else Fail)")
df['Result'] = df['Percentage'].apply(lambda x: 'Pass' if x >= 70 else 'Fail')
print(df)
```
